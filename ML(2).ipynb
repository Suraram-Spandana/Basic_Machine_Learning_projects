```python
import pandas as pd
from surprise import Dataset, Reader, SVD
from surprise.model_selection import train_test_split
```

```python
ratings = pd.read_csv("ratings.csv")
ratings.head()
```

```python
# 1. Define the reader with the rating scale (min rating is 0.5, max rating is 5.0)
reader = Reader(rating_scale=(0.5, 5.0))

# 2. Load the dataset from the pandas DataFrame containing columns: user, item, and rating
data = Dataset.load_from_df(ratings[['userId', 'movieId', 'rating']], reader)
```

```python
# Split the dataset into trainset (75%) and testset (25%) for training and evaluation
trainset, testset = train_test_split(data, test_size=0.25)
```

```python
# 1. Initialize the SVD (Singular Value Decomposition) recommendation algorithm
algo = SVD()

# 2. Train the algorithm on our training dataset to learn user-item preferences
algo.fit(trainset)
```

```python
# 1. Use the trained SVD algorithm to predict the rating for User ID 1 on Movie ID 302
pred = algo.predict(uid=1, iid=302)

# 2. Extract and print the estimated rating (.est) from the prediction object
print(f"Estimated Rating: {pred.est:.4f}")
```

```python
def recommend_for_user(user_id, num_recommendations=10):
    # 1. Get a list of all unique movie IDs in our dataset
    all_movies = ratings['movieId'].unique()

    # 2. Predict estimated ratings for this user across every movie
    predictions = [algo.predict(user_id, movie_id) for movie_id in all_movies]

    # 3. Sort predictions in descending order of their estimated rating (.est)
    predictions.sort(key=lambda x: x.est, reverse=True)

    # 4. Extract the item ID (movie ID) for the top N highest-rated predictions
    top_movies = [pred.iid for pred in predictions[:num_recommendations]]

    return top_movies
```

```python
print(recommend_for_user(1, 10))
```
